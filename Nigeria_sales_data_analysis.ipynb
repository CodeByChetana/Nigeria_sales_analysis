```python
df["Customer Name"].unique()
```

```python
df["Customer Name"]=df["Customer Name"].str.lower()
```

```python
df['Customer Name'].str.lower().value_counts()
```

```python
df["Customer Name"].fillna("Unknown",inplace=True)
```

```python
print(df['Customer Name'])
```

```python
df["State"].unique()
```

```python
df["State"]=df["State"].str.lower()
print(df["State"].unique())
```

```python
df["Product"].unique()
```

```python
df["Product"]=df["Product"].str.lower()
print(df["Product"].unique())
```

```python
df["Units Sold"].unique()
```

```python
print(df["Units Sold"].isnull().sum())
```

```python
df["Units Sold"]=df["Units Sold"].fillna(df['Total Sale']/df['Unit Price'])
print(df["Units Sold"])
```

```python
print(df["Units Sold"].isnull().sum())
```

```python
print(df["Unit Price"].isnull().sum())
```

```python
df["Unit Price"]=df["Unit Price"].fillna(df["Total Sale"]/df["Units Sold"])
print(df["Unit Price"].isnull().sum())
```

```python
df["Total Sale"]=df["Total Sale"].fillna(df["Units Sold"]*df["Unit Price"])
print(df["Total Sale"].isnull().sum())
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=["Units Sold", "Unit Price", "Total Sale"], how="all")
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=["Units Sold", "Total Sale"])
```

```python
df.isnull().sum()
```

```python
df["Unit Price"]=df["Unit Price"].fillna(df["Unit Price"].median())
print(df["Unit Price"])
```

```python
df["Sale Date"].unique()
```

```python
df["Sale Date"]=pd.to_datetime(df["Sale Date"])
print(df["Sale Date"].unique())
```

```python
df["Sales Channel"].unique()
```

```python
df["Sales Channel"]=df["Sales Channel"].fillna("Unknown")
```

```python
df["Sales Channel"].value_counts()
```

```python
df["Order ID"].unique()
```

```python
df["Order ID"]=df["Order ID"].fillna("Unknown")
print(df["Order ID"].isnull().sum())
```
